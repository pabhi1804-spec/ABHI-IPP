## 1. Display flower.jpg in a window titled Original Image.

```python
import cv2
from matplotlib import pyplot as plt
from PIL import Image
import numpy as np
from IPython.display import display
img = cv2.imread("flower.jpg")
if img is None: raise ValueError("Image not found or unsupported format")
cv2.imshow("Original Image", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

## 2. Print dimensions and data type of scene.jpg.

```python
img = cv2.imread("scene.jpg")
height, width = img.shape[:2]
channels = 1 if img.ndim == 2 else img.shape[2]
print("Height:", height, "Width:", width, "Channels:", channels)
print("Data type:", img.dtype)
```

## 3. Convert a colour image to grayscale and save output_gray.jpg.

```python
img = cv2.imread("flower.jpg")
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
cv2.imwrite("output_gray.jpg", gray)
print("Saved output_gray.jpg")
```

## 5. Resize an image to 50% width and height.

```python
img = cv2.imread("flower.jpg")
h, w = img.shape[:2]
small = cv2.resize(img, (int(w*0.5), int(h*0.5)))
plt.imshow(cv2.cvtColor(small, cv2.COLOR_BGR2RGB))
plt.title("Image resized to 50%"); plt.axis("off"); plt.show()
```

## 6. Crop the central 200×200 pixels and display with Matplotlib.

```python
img = cv2.imread("flower.jpg")
h, w = img.shape[:2]
if h < 200 or w < 200: raise ValueError("Image must be at least 200×200")
y, x = (h-200)//2, (w-200)//2
center_crop = img[y:y+200, x:x+200]
plt.imshow(cv2.cvtColor(center_crop, cv2.COLOR_BGR2RGB))
plt.title("Central 200×200 crop"); plt.axis("off"); plt.show()
```

## 7. Rotate an image 90 degrees clockwise.

```python
img = cv2.imread("flower.jpg")
rotated = cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE)
plt.imshow(cv2.cvtColor(rotated, cv2.COLOR_BGR2RGB))
plt.title("90° clockwise"); plt.axis("off"); plt.show()
```

## 8. Flip car.jpg left-to-right.

```python
img = cv2.imread("car.jpg")
flipped = cv2.flip(img, 1)  # flipCode=1 means horizontal
plt.imshow(cv2.cvtColor(flipped, cv2.COLOR_BGR2RGB))
plt.title("Horizontal flip"); plt.axis("off"); plt.show()
```

## 9. Convert RGB to HSV and display H, S and V channels.

```python
bgr = cv2.imread("flower.jpg")
rgb = cv2.cvtColor(bgr, cv2.COLOR_BGR2RGB)
hsv = cv2.cvtColor(rgb, cv2.COLOR_RGB2HSV)
h, s, v = cv2.split(hsv)
fig, axes = plt.subplots(1, 3, figsize=(12,4))
for ax, channel, title in zip(axes, [h,s,v], ["Hue","Saturation","Value"]):
    ax.imshow(channel, cmap="gray"); ax.set_title(title); ax.axis("off")
plt.tight_layout(); plt.show()
```

## 10. Calculate MSE between img1.jpg and img2.jpg.

```python
a = cv2.imread("img1.jpg"), cv2.IMREAD_GRAYSCALE
b = cv2.imread("img2.jpg"), cv2.IMREAD_GRAYSCALE
if a.shape != b.shape: b = cv2.resize(b, (a.shape[1], a.shape[0]))
mse = np.mean((a.astype(np.float32)-b.astype(np.float32))**2)
print("MSE:", mse)
```
